```python
import geopandas as gpd
```

```python
charging_stations = gpd.read_file(r'../Data/ev_charging_units.geojson')
charging_stations.head(3)
```

```python
cleaned = charging_stations[['City', 'Station Name', 'EV J3400 Power Output (kW)', 'EV J3400 Connector Count', 'geometry']]
cleaned
```

```python
cleaned = cleaned[cleaned['City'] == 'Seattle']
cleaned
```

```python
import pandas as pd

cleaned['EV J3400 Connector Count'] = pd.to_numeric(
    cleaned['EV J3400 Connector Count'], errors='coerce'
)
```

```python
grouped = (
    cleaned.groupby('Station Name', as_index=False)
    .agg({
        'EV J3400 Connector Count': 'sum',
        'City': 'first',
        'geometry': 'first'
        
    })
)
grouped
```

```python
grouped = gpd.GeoDataFrame(grouped, geometry='geometry', crs=cleaned.crs)
grouped.to_file("grouped_ev_stations.geojson", driver="GeoJSON")
```
